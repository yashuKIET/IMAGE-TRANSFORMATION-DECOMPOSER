```python
# ============================================================
# IMAGE TRANSFORMATION DECOMPOSER
# FRONT-END + ANIMATION
# GOOGLE COLAB - SINGLE PYTHON CELL
# ============================================================

from IPython.display import HTML, display

html_code = r"""
<!DOCTYPE html>
<html>
<head>

<meta charset="UTF-8">

<style>

* {
    box-sizing: border-box;
}

body {
    margin: 0;
    font-family: Arial, sans-serif;
    background: #0f172a;
    color: white;
}

.container {
    max-width: 1100px;
    margin: auto;
    padding: 25px;
}

h1 {
    text-align: center;
    color: #38bdf8;
    margin-bottom: 5px;
}

.subtitle {
    text-align: center;
    color: #94a3b8;
    margin-bottom: 25px;
}

.panel {
    background: #1e293b;
    padding: 20px;
    border-radius: 15px;
    margin-bottom: 20px;
}

.panel h2 {
    color: #38bdf8;
    margin-top: 0;
}

.matrix {
    display: grid;
    grid-template-columns: 100px 100px;
    gap: 12px;
    justify-content: center;
    margin: 20px;
}

.matrix input {
    width: 100px;
    height: 55px;
    text-align: center;
    font-size: 20px;
    border-radius: 10px;
    border: 2px solid #38bdf8;
    background: #0f172a;
    color: white;
}

.buttons {
    display: flex;
    justify-content: center;
    gap: 12px;
    flex-wrap: wrap;
}

button {
    padding: 12px 20px;
    border: none;
    border-radius: 10px;
    font-size: 16px;
    font-weight: bold;
    cursor: pointer;
}

.animate {
    background: #22c55e;
    color: white;
}

.pause {
    background: #f59e0b;
    color: white;
}

.reset {
    background: #ef4444;
    color: white;
}

button:hover {
    opacity: 0.85;
}

.results {
    display: grid;
    grid-template-columns: repeat(2, 1fr);
    gap: 15px;
}

.result {
    background: #0f172a;
    padding: 15px;
    border-radius: 10px;
}

.result h3 {
    color: #38bdf8;
}

#canvas {
    width: 100%;
    max-width: 850px;
    height: 500px;
    display: block;
    margin: auto;
    background: white;
    border-radius: 12px;
}

.status {
    text-align: center;
    margin-top: 12px;
    color: #94a3b8;
}

.formula {
    background: #020617;
    padding: 15px;
    border-radius: 10px;
    line-height: 1.8;
}

@media(max-width:700px) {

    .results {
        grid-template-columns: 1fr;
    }

    #canvas {
        height: 350px;
    }

}

</style>

</head>

<body>

<div class="container">

<h1>🧮 IMAGE TRANSFORMATION DECOMPOSER</h1>

<div class="subtitle">
B10 • Eigenvalues • Eigenvectors • Rotation • Stretching
</div>


<!-- MATRIX INPUT -->

<div class="panel">

<h2>1️⃣ Enter 2×2 Transformation Matrix</h2>

<div class="matrix">

<input id="a" type="number" value="2" step="0.1">

<input id="b" type="number" value="1" step="0.1">

<input id="c" type="number" value="1" step="0.1">

<input id="d" type="number" value="2" step="0.1">

</div>

<div class="formula">

A = [ a&nbsp;&nbsp;b ]<br>
&nbsp;&nbsp;&nbsp;&nbsp;[ c&nbsp;&nbsp;d ]

<br><br>

Transformation:

<br>

x′ = A × x

</div>

<br>

<div class="buttons">

<button class="animate" id="animateBtn">
▶ Animate
</button>

<button class="pause" id="pauseBtn">
⏸ Pause
</button>

<button class="reset" id="resetBtn">
🔄 Reset
</button>

</div>

<div class="status" id="status">
Ready
</div>

</div>


<!-- RESULTS -->

<div class="results">

<div class="panel result">

<h3>📐 Matrix</h3>

<div id="matrixResult"></div>

</div>


<div class="panel result">

<h3>🔢 Eigenvalues</h3>

<div id="eigenResult"></div>

</div>


<div class="panel result">

<h3>🧭 Eigenvectors</h3>

<div id="vectorResult"></div>

</div>


<div class="panel result">

<h3>🔄 Rotation + Stretching</h3>

<div id="decomposition"></div>

</div>

</div>


<!-- ANIMATION -->

<div class="panel">

<h2>2️⃣ Animated Transformation</h2>

<canvas
id="canvas"
width="850"
height="500">
</canvas>

<div class="status">

🔵 Original
&nbsp;&nbsp; 🔴 Transformed
&nbsp;&nbsp; 🟢🟣 Eigenvector Directions

</div>

</div>


<!-- MATHEMATICS -->

<div class="panel">

<h2>3️⃣ Mathematical Concept</h2>

<div class="formula">

<b>Characteristic Equation</b>

<br>

det(A − λI) = 0

<br><br>

<b>Eigenvector Equation</b>

<br>

Av = λv

<br><br>

<b>Eigenvector</b>

→ Direction that remains unchanged.

<br>

<b>Eigenvalue</b>

→ Scaling / stretching factor.

<br>

<b>Polar Decomposition</b>

<br>

A = R × S

<br>

R → Rotation / Reflection

<br>

S → Stretching / Compression

</div>

</div>

</div>


<script>

/* =========================================================
   ELEMENTS
========================================================= */

const canvas = document.getElementById("canvas");

const ctx = canvas.getContext("2d");

const aInput = document.getElementById("a");
const bInput = document.getElementById("b");
const cInput = document.getElementById("c");
const dInput = document.getElementById("d");

const matrixResult =
    document.getElementById("matrixResult");

const eigenResult =
    document.getElementById("eigenResult");

const vectorResult =
    document.getElementById("vectorResult");

const decomposition =
    document.getElementById("decomposition");

const status =
    document.getElementById("status");


let running = false;

let progress = 0;

let animationFrame = null;


/* =========================================================
   GET MATRIX
========================================================= */

function getMatrix() {

    return {

        a: Number(aInput.value) || 0,

        b: Number(bInput.value) || 0,

        c: Number(cInput.value) || 0,

        d: Number(dInput.value) || 0

    };

}


/* =========================================================
   EIGENVALUES
========================================================= */

function getEigenvalues(A) {

    const trace = A.a + A.d;

    const determinant =
        A.a * A.d -
        A.b * A.c;

    const discriminant =
        trace * trace -
        4 * determinant;


    if (discriminant < 0) {

        return null;

    }


    const root =
        Math.sqrt(discriminant);


    return [

        (trace + root) / 2,

        (trace - root) / 2

    ];

}


/* =========================================================
   EIGENVECTOR
========================================================= */

function getEigenvector(A, lambda) {

    let x;
    let y;


    if (Math.abs(A.b) > 0.000001) {

        x = 1;

        y =
            (lambda - A.a) / A.b;

    }

    else if (Math.abs(A.c) > 0.000001) {

        y = 1;

        x =
            (lambda - A.d) / A.c;

    }

    else {

        if (
            Math.abs(A.a - lambda)
            < 0.000001
        ) {

            x = 1;
            y = 0;

        }

        else {

            x = 0;
            y = 1;

        }

    }


    const length =
        Math.sqrt(x * x + y * y);


    if (length === 0) {

        return {
            x: 1,
            y: 0
        };

    }


    return {

        x: x / length,

        y: y / length

    };

}


/* =========================================================
   UPDATE INFORMATION
========================================================= */

function updateResults() {

    const A = getMatrix();


    matrixResult.innerHTML =

        "A = [ "
        + A.a.toFixed(2)
        + " &nbsp; "
        + A.b.toFixed(2)
        + " ]<br>"

        +

        "&nbsp;&nbsp;&nbsp;&nbsp;[ "
        + A.c.toFixed(2)
        + " &nbsp; "
        + A.d.toFixed(2)
        + " ]";


    const eigenvalues =
        getEigenvalues(A);


    if (eigenvalues === null) {

        eigenResult.innerHTML =
            "Complex eigenvalues";

        vectorResult.innerHTML =
            "No real eigenvectors";

        decomposition.innerHTML =
            "Real eigenvector visualization unavailable.";

        return;

    }


    const lambda1 =
        eigenvalues[0];

    const lambda2 =
        eigenvalues[1];


    eigenResult.innerHTML =

        "λ₁ = "
        + lambda1.toFixed(4)
        + "<br>"

        +

        "λ₂ = "
        + lambda2.toFixed(4);


    const v1 =
        getEigenvector(
            A,
            lambda1
        );


    const v2 =
        getEigenvector(
            A,
            lambda2
        );


    vectorResult.innerHTML =

        "v₁ = ("
        + v1.x.toFixed(3)
        + ", "
        + v1.y.toFixed(3)
        + ")<br>"

        +

        "v₂ = ("
        + v2.x.toFixed(3)
        + ", "
        + v2.y.toFixed(3)
        + ")";


    const angle =
        Math.atan2(
            A.c - A.b,
            A.a + A.d
        )
        * 180
        / Math.PI;


    decomposition.innerHTML =

        "Approx. rotation angle = "
        + angle.toFixed(2)
        + "°<br><br>"

        +

        "Stretching factors:<br>"

        +

        "λ₁ = "
        + lambda1.toFixed(3)
        + "<br>"

        +

        "λ₂ = "
        + lambda2.toFixed(3);


    drawScene(0);

}


/* =========================================================
   DRAW GRID
========================================================= */

function drawGrid() {

    ctx.strokeStyle =
        "#e2e8f0";

    ctx.lineWidth = 1;


    for (
        let x = 0;
        x <= canvas.width;
        x += 40
    ) {

        ctx.beginPath();

        ctx.moveTo(x, 0);

        ctx.lineTo(
            x,
            canvas.height
        );

        ctx.stroke();

    }


    for (
        let y = 0;
        y <= canvas.height;
        y += 40
    ) {

        ctx.beginPath();

        ctx.moveTo(0, y);

        ctx.lineTo(
            canvas.width,
            y
        );

        ctx.stroke();

    }


    ctx.strokeStyle =
        "#64748b";

    ctx.lineWidth = 2;


    /* X axis */

    ctx.beginPath();

    ctx.moveTo(
        0,
        250
    );

    ctx.lineTo(
        850,
        250
    );

    ctx.stroke();


    /* Y axis */

    ctx.beginPath();

    ctx.moveTo(
        425,
        0
    );

    ctx.lineTo(
        425,
        500
    );

    ctx.stroke();

}


/* =========================================================
   TRANSFORM POINT
========================================================= */

function transformPoint(x, y, A) {

    return {

        x:
            A.a * x +
            A.b * y,

        y:
            A.c * x +
            A.d * y

    };

}


/* =========================================================
   DRAW POLYGON
========================================================= */

function drawPolygon(points, color) {

    ctx.strokeStyle = color;

    ctx.lineWidth = 4;

    ctx.beginPath();


    points.forEach(
        function(point, index) {

            const sx =
                425 +
                point.x * 60;


            const sy =
                250 -
                point.y * 60;


            if (index === 0) {

                ctx.moveTo(
                    sx,
                    sy
                );

            }

            else {

                ctx.lineTo(
                    sx,
                    sy
                );

            }

        }
    );


    ctx.closePath();

    ctx.stroke();

}


/* =========================================================
   DRAW EIGENVECTORS
========================================================= */

function drawEigenvectors() {

    const A = getMatrix();

    const eigenvalues =
        getEigenvalues(A);


    if (eigenvalues === null) {

        return;

    }


    const v1 =
        getEigenvector(
            A,
            eigenvalues[0]
        );


    const v2 =
        getEigenvector(
            A,
            eigenvalues[1]
        );


    /* Eigenvector 1 */

    ctx.strokeStyle =
        "#22c55e";

    ctx.lineWidth = 3;

    ctx.beginPath();

    ctx.moveTo(

        425 - v1.x * 180,

        250 + v1.y * 180

    );

    ctx.lineTo(

        425 + v1.x * 180,

        250 - v1.y * 180

    );

    ctx.stroke();


    /* Eigenvector 2 */

    ctx.strokeStyle =
        "#a855f7";

    ctx.beginPath();

    ctx.moveTo(

        425 - v2.x * 180,

        250 + v2.y * 180

    );

    ctx.lineTo(

        425 + v2.x * 180,

        250 - v2.y * 180

    );

    ctx.stroke();

}


/* =========================================================
   MAIN DRAW
========================================================= */

function drawScene(t) {

    ctx.clearRect(
        0,
        0,
        canvas.width,
        canvas.height
    );


    drawGrid();


    const A = getMatrix();


    const original = [

        { x: -1, y: -1 },

        { x: 1, y: -1 },

        { x: 1, y: 1 },

        { x: -1, y: 1 }

    ];


    const transformed =
        original.map(
            function(point) {

                return transformPoint(
                    point.x,
                    point.y,
                    A
                );

            }
        );


    const animated =
        original.map(
            function(point, index) {

                return {

                    x:
                        point.x
                        +
                        (
                            transformed[index].x
                            -
                            point.x
                        )
                        * t,

                    y:
                        point.y
                        +
                        (
                            transformed[index].y
                            -
                            point.y
                        )
                        * t

                };

            }
        );


    /* Original shape */

    if (t < 1) {

        drawPolygon(
            original,
            "#2563eb"
        );

    }


    /* Animated shape */

    drawPolygon(
        animated,
        "#ef4444"
    );


    /* Eigenvectors */

    drawEigenvectors();


    /* Percentage */

    ctx.fillStyle =
        "#0f172a";

    ctx.font =
        "bold 18px Arial";


    ctx.fillText(

        "Transformation: "
        +
        Math.round(t * 100)
        +
        "%",

        20,
        30

    );

}


/* =========================================================
   ANIMATION
========================================================= */

function animate() {

    running = true;

    const duration = 2500;

    const start =
        performance.now()
        -
        progress * duration;


    function frame(time) {

        if (!running) {

            return;

        }


        progress =
            Math.min(
                1,
                (time - start)
                / duration
            );


        drawScene(progress);


        if (progress < 1) {

            animationFrame =
                requestAnimationFrame(
                    frame
                );

        }

        else {

            running = false;

            status.innerText =
                "✅ Transformation complete";

        }

    }


    animationFrame =
        requestAnimationFrame(
            frame
        );

}


/* =========================================================
   ANIMATE BUTTON
========================================================= */

document
    .getElementById("animateBtn")
    .addEventListener(
        "click",
        function() {

            running = false;

            cancelAnimationFrame(
                animationFrame
            );

            updateResults();

            progress = 0;

            status.innerText =
                "🎬 Animation running...";

            animate();

        }
    );


/* =========================================================
   PAUSE
========================================================= */

document
    .getElementById("pauseBtn")
    .addEventListener(
        "click",
        function() {

            running = false;

            cancelAnimationFrame(
                animationFrame
            );

            status.innerText =
                "⏸ Animation paused";

        }
    );


/* =========================================================
   RESET
========================================================= */

document
    .getElementById("resetBtn")
    .addEventListener(
        "click",
        function() {

            running = false;

            cancelAnimationFrame(
                animationFrame
            );

            progress = 0;

            updateResults();

            drawScene(0);

            status.innerText =
                "🔄 Reset complete";

        }
    );


/* =========================================================
   LIVE MATRIX INPUT
========================================================= */

[
    aInput,
    bInput,
    cInput,
    dInput
].forEach(
    function(input) {

        input.addEventListener(
            "input",
            function() {

                running = false;

                cancelAnimationFrame(
                    animationFrame
                );

                progress = 0;

                updateResults();

                status.innerText =
                    "Matrix updated";

            }
        );

    }
);


/* =========================================================
   START
========================================================= */

updateResults();

drawScene(0);

</script>

</body>
</html>
"""

display(HTML(html_code))
```
